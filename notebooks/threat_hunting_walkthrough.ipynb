# Threat Hunting Walkthrough

This notebook provides a simple walkthrough of failed logins and suspicious authentication patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/sample_auth_logs.csv')
df.head()
```

```python
df['status'].value_counts()
```

```python
failed_by_ip = df[df['status'] == 'failed'].groupby('source_ip').size().sort_values(ascending=False)
failed_by_ip
```

```python
failed_by_ip.plot(kind='bar', figsize=(8, 4), color='#c1121f')
plt.title('Failed Logins by Source IP')
plt.ylabel('Count')
plt.show()
```
